## Завдання 4. Одновимірний масив NumPy (перші 10 парних чисел)

```python
even = np.arange(2, 21, 2)   # 2, 4, ..., 20
print("Масив:", even)
print("Сума:", even.sum())
print("Середнє:", even.mean())
print("Мінімум:", even.min())
print("Максимум:", even.max())
print("Масив * 3:", even * 3)
```

## Завдання 5. Двовимірний масив 3×3 з випадковими цілими від 1 до 10

```python
rng = np.random.default_rng()
m = rng.integers(1, 11, size=(3, 3))   # верхня межа не включається, тому 11
print(m)
print("Сума за стовпцями (axis=0):", m.sum(axis=0))
print("Сума за рядками (axis=1):", m.sum(axis=1))
```
